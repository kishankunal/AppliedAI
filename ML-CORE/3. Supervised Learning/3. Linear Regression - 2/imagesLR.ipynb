```python
"""
Step-by-step: Age Prediction from Face Images using Linear and Logistic Regression

- Loads female face images from training and test folders.
- Extracts the age label from the filename.
- Preprocesses images with skimage (not keras).
- Builds feature and label arrays for regression/classification.
- Visualizes a sample of the data.
- Trains linear regression (predict exact age) and logistic regression (predict age group).
"""

import numpy as np
import os
from skimage import io, color, transform
import matplotlib.pyplot as plt
from glob import glob

# --- Helper functions ---

def load_images_labels(folder, max_images=200, female_only=True):
    """
    Loads up to `max_images` grayscale images and their age labels from a folder.
    Only loads female images (gender=1) if female_only=True.
    Images are resized and flattened into vectors.
    """
    X = []
    y = []
    image_dict = {}

    files = sorted(glob(os.path.join(folder, "*.jpg")))
    count = 0

    for filepath in files:
        # Filename format: [age]_[gender]_[race]_[date&time].jpg
        fname = os.path.basename(filepath)
        parts = fname.split("_")
        if len(parts) < 4:
            continue
        age = int(parts[0])
        gender = int(parts[1])

        if female_only and gender != 1:
            continue  # skip non-female faces

        # Read and process image
        img = io.imread(filepath)
        if img.ndim == 3:
            img = color.rgb2gray(img)
        img_resized = transform.resize(img, (200, 200), anti_aliasing=True)
        X.append(img_resized.flatten())
        y.append(age)
        image_dict[count] = img_resized
        count += 1
        if count >= max_images:
            break
    return np.array(X), np.array(y), image_dict

# --- Load data ---
train_dir = "training_images"
test_dir = "testing_images"

# For demonstration, load 200 training and 50 testing images
X_train, y_train, train_images = load_images_labels(train_dir, max_images=200)
X_test, y_test, test_images = load_images_labels(test_dir, max_images=50)

print("Shape of training data:", X_train.shape)
print("Example ages in training set:", y_train[:10])

# --- Display a sample grid of images with ages ---
plt.figure(figsize=(10, 10))
for i in range(20):
    plt.subplot(4, 5, i+1)
    plt.imshow(train_images[i], cmap="gray")
    plt.title(f"Age: {y_train[i]}")
    plt.axis("off")
plt.tight_layout()
plt.show()


```

```python
# --- Display a sample grid of images with ages ---
plt.figure(figsize=(10, 10))
for i in range(20):
    plt.subplot(4, 5, i+1)
    plt.imshow(test_images[i], cmap="gray")
    plt.title(f"Age: {y_test[i]}")
    plt.axis("off")
plt.tight_layout()
plt.show()
```

```python
# --- Linear Regression: Predicting Age as a Continuous Value ---
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, r2_score, mean_squared_error

linreg = LinearRegression()
linreg.fit(X_train, y_train)
y_pred_lin = linreg.predict(X_test)

print("\n--- Linear Regression Results ---")
print("Mean Absolute Error (MAE):", mean_absolute_error(y_test, y_pred_lin))
print("R2 Score:", r2_score(y_test, y_pred_lin))

# Calculate RMSE
rmse = np.sqrt(mean_squared_error(y_test, y_pred_lin))
print("Root Mean Squared Error (RMSE):", rmse)
```

```python
import random
import matplotlib.pyplot as plt

# Pick 5 random indices from the test set
random.seed(42)
indices = random.sample(range(len(y_test)), 5)

plt.figure(figsize=(15, 5))
for i, idx in enumerate(indices):
    plt.subplot(1, 5, i+1)
    # Show the image
    plt.imshow(test_images[idx], cmap='gray')
    plt.axis('off')
    # Show true and predicted age in title
    true_age = y_test[idx]
    pred_age = int(round(y_pred_lin[idx]))
    plt.title(f"Actual: {true_age}\nPred: {pred_age}", fontsize=12)
plt.suptitle("Model Performance: 5 Random Test Images\n(Actual vs. Predicted Age)", fontsize=16)
plt.tight_layout(rect=[0, 0.03, 1, 0.85])
plt.show()
```



### **Normalization**

* **StandardScaler:** Each pixel’s distribution is centered and scaled (mean=0, std=1).
* **MinMaxScaler:** All pixel values scaled between 0 and 1.
* **Compares model performance** after each type of scaling.
* Only prints out the **RMSE** and **R²** for easy comparison.

---



```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np

results = {}

# Standard Scaler
scaler_std = StandardScaler()
X_train_std = scaler_std.fit_transform(X_train)
X_test_std = scaler_std.transform(X_test)

linreg_std = LinearRegression()
linreg_std.fit(X_train_std, y_train)
y_pred_std = linreg_std.predict(X_test_std)

rmse_std = np.sqrt(mean_squared_error(y_test, y_pred_std))
r2_std = r2_score(y_test, y_pred_std)
results['StandardScaler'] = (rmse_std, r2_std)

# MinMax Scaler
scaler_mm = MinMaxScaler()
X_train_mm = scaler_mm.fit_transform(X_train)
X_test_mm = scaler_mm.transform(X_test)

linreg_mm = LinearRegression()
linreg_mm.fit(X_train_mm, y_train)
y_pred_mm = linreg_mm.predict(X_test_mm)

rmse_mm = np.sqrt(mean_squared_error(y_test, y_pred_mm))
r2_mm = r2_score(y_test, y_pred_mm)
results['MinMaxScaler'] = (rmse_mm, r2_mm)

# Display results
for method, (rmse, r2) in results.items():
    print(f"{method}:")
    print(f"  RMSE = {rmse:.2f}")
    print(f"  R2   = {r2:.4f}\n")
```
